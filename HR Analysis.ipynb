# KPI 1. Overall Attrition Rate 

```python
df.columns
```

```python
attrition_counts = df["Attrition"].value_counts()
print("attrition_counts")
attrition_rate = (attrition_counts["Yes"]/len(df))*100
print (f"Attrition Rate:{attrition_rate:.4f}%")
```

# KPI 2. Attrition Rate by Department

```python
dept_attrition = df.groupby('Department')['Attrition'].value_counts(normalize=True) * 100
print(dept_attrition)
```

```python
dept_Summery = df.groupby('Department')['Attrition'].value_counts().unstack()
print(dept_Summery)
```

```python
dept_summary = df.groupby('Department')['Attrition'].value_counts().unstack()
dept_summary['Total'] = dept_summary['No'] + dept_summary['Yes']
dept_summary['Attrition_Rate_%'] = (dept_summary['Yes'] / dept_summary['Total']) * 100

print(dept_summary)
```

# KPI 3. Attrition Rate by OverTime

```python
overtime_summary = df.groupby('OverTime')['Attrition'].value_counts().unstack()
overtime_summary['Total'] = overtime_summary['No'] + overtime_summary['Yes']
overtime_summary['Attrition_Rate_%'] = (overtime_summary['Yes'] / overtime_summary['Total'] * 100).round(2)

print(overtime_summary)
```

# KPI 4. Attrition by Department + OverTime Combined

```python
combined_summary = df.groupby(['Department', 'OverTime'])['Attrition'].value_counts().unstack()
combined_summary['Total'] = combined_summary['No'] + combined_summary['Yes']
combined_summary['Attrition_Rate_%'] = (combined_summary['Yes'] / combined_summary['Total'] * 100).round(2)

print(combined_summary)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Reset index so Department and OverTime become normal columns again
plot_data = combined_summary.reset_index()

plt.figure(figsize=(8,5))
sns.barplot(data=plot_data, x='Department', y='Attrition_Rate_%', hue='OverTime')
plt.title('Attrition Rate by Department and OverTime')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=15)
plt.show()
```

# KPI 5. Attrition Rate by Job Role

```python
role_summary = df.groupby('JobRole')['Attrition'].value_counts().unstack()
role_summary['Total'] = role_summary['No'] + role_summary['Yes']
role_summary['Attrition_Rate_%'] = (role_summary['Yes'] / role_summary['Total'] * 100).round(2)

print(role_summary)
```

```python

plot_data = role_summary.sort_values('Attrition_Rate_%', ascending=False).reset_index()

plt.figure(figsize=(8,6))
ax = sns.barplot(data=plot_data, y='JobRole', x='Attrition_Rate_%', color='#eb6834')

# Add the number label on top of (at the end of) each bar
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', padding=3)

plt.title('Attrition Rate by Job Role')
plt.xlabel('Attrition Rate (%)')
plt.ylabel('')
plt.xlim(0, plot_data['Attrition_Rate_%'].max() + 5)  # extra space so labels don't get cut off
plt.show()
```
